```
# 2.2 Write a function `calculate_average_temperature(temp_data_list)`


def calculate_average_temperature(temp_data_list):
    """
    Calculates the average temperature from a list of dictionaries.
    Returns 0.0 if the list is empty.
    """
    if not temp_data_list:
        return 0.0


    total_temp = sum(item['temperature_celsius'] for item in temp_data_list)


    return total_temp / len(temp_data_list)



avg_temp = calculate_average_temperature(temperature_readings)
print(f"Calculated average temperature: {avg_temp:.2s}°C" if isinstance(avg_temp, str) else f"Calculated average temperature: {avg_temp:.2f}°C")
```

```
import datetime
import random

DEVICE_TYPES = ['Temperature_Sensor', 'Humidity_Sensor', 'Pressure_Sensor', 'Light_Sensor']
LOCATIONS = ['Lab1', 'Lab2', 'OfficeA', 'OfficeB', 'Warehouse']

READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0),
    'Humidity_Sensor': (30.0, 80.0),
    'Pressure_Sensor': (980.0, 1030.0),
    'Light_Sensor': (100, 1000)
}

raw_sensor_data = []

for i in range(1, 101):
    device_type = random.choice(DEVICE_TYPES)
    value = round(random.uniform(*READING_RANGES[device_type]), 2)

    raw_sensor_data.append((
        datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        f"DEV{i:03d}",
        device_type,
        random.choice(LOCATIONS),
        value
    ))

print(raw_sensor_data[:10])
print("Total Readings:", len(raw_sensor_data))
```

```
# 1.4 Create a dictionary named 'device_stats' where keys are 'device_type' and values are

temp_stats = {}

for item in raw_sensor_data:
    device_type = item[2]
    value = item[4]

    if device_type not in temp_stats:
        temp_stats[device_type] = [float('inf'), float('-inf')]


    if value < temp_stats[device_type][0]:
        temp_stats[device_type][0] = value
    if value > temp_stats[device_type][1]:
        temp_stats[device_type][1] = value


device_stats = {d_type: (limits[0], limits[1]) for d_type, limits in temp_stats.items()}

print("\nDevice Statistics (Min, Max):")
for d_type, stats in device_stats.items():
    print(f"  {d_type}: {stats}")
```

```
import os
import datetime
import random

print("--- Intermediate Python Core Exercise ---")
print("This exercise covers Python fundamentals, data structures, functions,")
print("core libraries (os, datetime, random), file handling, and exception handling.")
print("Estimated completion time: 3-4 hours.")
print("------------------------------------------")

print("\n--- Task 1: Data Generation and Basic Structures ---")



locations = ["Lab_A", "Warehouse_B", "Server_Room_C"]
sensor_types = ["Temperature_Sensor", "Humidity_Sensor", "Pressure_Sensor"]
base_time = datetime.datetime.now()

raw_sensor_data = []

for i in range(20):
    s_type = random.choice(sensor_types)
    val = round(random.uniform(18.0, 35.0), 2) if s_type == "Temperature_Sensor" else round(random.uniform(30.0, 60.0), 2)


    item_tuple = (
        (base_time - datetime.timedelta(minutes=i)).isoformat(),
        f"{s_type[:3].upper()}_{100 + i}",
        s_type,
        random.choice(locations),
        val
    )
    raw_sensor_data.append(item_tuple)



temperature_readings = [
    {
        'timestamp': timestamp,
        'device_id': device_id,
        'location': location,
        'temperature_celsius': value
    }
    for timestamp, device_id, device_type, location, value in raw_sensor_data
    if device_type == 'Temperature_Sensor'
]


print(f"Total raw records generated: {len(raw_sensor_data)}")
print(f"Extracted {len(temperature_readings)} temperature readings.")

if temperature_readings:
    print("\nSample processed dictionary output:")
    print(temperature_readings[0])
```
