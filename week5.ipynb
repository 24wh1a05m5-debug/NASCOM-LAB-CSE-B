```python
import numpy as np
import pandas as pd
from functools import reduce

# Create a NumPy array
array = np.array([[1, 2], [3, 4], [5, 6]])

# Create a DataFrame from the NumPy array
df_from_array = pd.DataFrame(array, columns=['Column1', 'Column2'])

# Get numbers from the user
numbers = list(map(int, input("Enter the list of elements: ").split()))

# Square each number
squared = list(map(lambda x: x ** 2, numbers))

# Filter even squares
even_numbers = list(filter(lambda x: x % 2 == 0, squared))

# Find the sum of even squares
sum_even = reduce(lambda x, y: x + y, even_numbers, 0)

# Display results
print("\nOriginal List :", numbers)
print("Squared List  :", squared)
print("Even Squares  :", even_numbers)
print("Sum of Even Squares :", sum_even)

# Display DataFrame
print("\nDataFrame from NumPy array:")
print(df_from_array)



```

```python

import numpy as np
import pandas as pd
np.random.seed(1)
num_rows = 10
df_synth = pd.DataFrame({
    'ID': np.arange(1, num_rows + 1),
    'Score': np.random.randint(50, 100, size=num_rows),
    'Passed': np.random.choice([True, False], size=num_rows)
})

print("\nSynthetic DataFrame:")
print(df_synth)
```

```python

print("\nStudents who passed:")
print(df_synth[df_synth['Passed'] == True])

print("\nAverage score of failed students:")
print(df_synth[df_synth['Passed'] == False]['Score'].mean())
```
